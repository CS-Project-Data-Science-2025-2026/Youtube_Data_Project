## 1- Create a single dataframe with the concatenation of all input csv files, adding a column called country

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Path for connect to CSV files location
data_path = Path("/Users/tonydawra/Downloads/projectfocstry/datasets")

# Abbreviations for each file according to file names
country_list = ['CA','DE','FR','GB','IN','JP','KR','MX','RU','US']

#  This list for combine each country's DataFrame
all_files = []

# #Loop for reading every CSV file based on country code and add a "country" column
for country in country_list:
    # Reaching each file based on country name
    file_path = data_path / f"{country}videos.csv"
    
    # Read the CSV; ignore errors because of "RUvideos", "JPvideos","USvideos" files
    df_country = pd.read_csv(file_path, encoding='utf-8', encoding_errors="ignore")
    
    # Creating new column for country names in order to concatenate 
    df_country["country"] = country
    
    # Store this temporary DataFrame in the list
    all_files.append(df_country)

# Concatenate all DataFrames by using country column
df = pd.concat(all_files, ignore_index=True)
```

## 2- Delete no tags

```python
# Keep only rows where the "tags" column is different from the literal string "[none]"
df = df[df["tags"] != "[none]"]
df.head()
```

## 3. For each channel, determine the total number of views

```python
# Group by channel title and sum the views across all its videos
df.groupby("channel_title")[["views"]].sum()
```

## 4-Save all rows with disabled comments and disabled ratings, or that have video_error_or_removed in a new dataframe called excluded, and remove those rows from the original dataframe.

```python
# Select the rows that meet one of these conditions:
#  - comments_disabled == True AND ratings_disabled == True
#  - video_error_or_removed == True
excluded = df[((df["comments_disabled"] == True) & (df["ratings_disabled"] == True)) | (df["video_error_or_removed"] == True)]
excluded
```

```python
# Now clean the main dataframe by *removing* those rows
df = df[((df["comments_disabled"] != True) & (df["ratings_disabled"] != True)) | (df["video_error_or_removed"] != True)]
df
```

## 5- Add a like_ratio column storing the ratio between the number of likes and of dislikes

```python
# Create a new column with the ratio of likes to dislikes
df["like_ratio"] = df["likes"] / df["dislikes"]
df.head()
```

```python
# just inspect the three relevant columns
df[["likes", "dislikes", "like_ratio"]].head()
```

## 6- Cluster the publish time into 10-minute intervals (e.g. from 02:20 to 02:30)

```python
# Convert the publish_time column to datetime; any invalid format becomes NaT
df["publish_time"] = pd.to_datetime(df["publish_time"], errors="coerce")

# "Floor" each publish_time to the previous 10-minute mark
df["publish_interval_10m"] = df["publish_time"].dt.floor("10min")

df[["publish_time", "publish_interval_10m"]].head()
```

## 7- For each interval, determine the number of videos, average number of likes and of dislikes.

```python
# Group by the 10-minute interval
grouped = df.groupby("publish_interval_10m")

# Compute mean likes and dislikes per interval
interval_stats = grouped[["likes", "dislikes"]].mean()

# Add a column with the number of videos in each interval
interval_stats["n_videos"] = grouped.size()

# Reset the index so that publish_interval_10m becomes a normal column
interval_stats = interval_stats.reset_index()
interval_stats
```

## 8- For each tag, determine the number of videos
Notice that tags contains a string with several tags.

```python
# Work on a copy to avoid accidentally modifying the original df
df_tags = df.copy()

# Explode the "tags" string into one row per tag:
#  1. Drop rows where "tags" is NaN
#  2. Split the "tags" string on "|" into a list
#  3. explode() creates one row per element of that list
df_tags_exploded = (
    df_tags
    .dropna(subset=["tags"])
    .assign(tag_list=lambda x: x["tags"].str.split("|"))
    .explode("tag_list")
)

# Clean the exploded tag: remove extra spaces and surrounding quotes
df_tags_exploded["tag"] = (
    df_tags_exploded["tag_list"]
    .str.strip()
    .str.strip('"')
)

# We no longer need the intermediate list column
df_tags_exploded = df_tags_exploded.drop(columns=["tag_list"])

# Inspect how tags look now
df_tags_exploded[["video_id", "tags", "tag", "country"]]
```

```python
# Count how many *distinct* videos contain each tag
videos_per_tag = (
    df_tags_exploded.groupby("tag")["video_id"]
    .nunique()                     # number of distinct videos per tag
    .reset_index(name="n_videos")
)
```

## 9- Find the tags with the largest number of videos

```python
# Sort tags by number of videos (descending) and look at the top 10
top_tags = videos_per_tag.sort_values("n_videos", ascending=False)
top_tags.head(10)
```

## 10- For each (tag, country) pair, compute average ratio likes/dislikes


```python
# Aggregate total likes, total dislikes, and number of rows (videos) per (tag, country)
df_sum_like_ratio = df_tags_exploded.groupby(["tag", "country"]).agg(
    {
        "likes": "sum",
        "dislikes": "sum",
        "video_id": "count"
    }
)

# Compute the like/dislike ratio using the aggregated sums
df_sum_like_ratio["like_ratio_tag_country"] = (
    df_sum_like_ratio["likes"] / df_sum_like_ratio["dislikes"]
)

# Replace infinite values (division by zero) with 0 and fill NaN with 0
df_sum_like_ratio["like_ratio_tag_country"] = (
    df_sum_like_ratio["like_ratio_tag_country"]
    .replace([np.inf, -np.inf], 0)
    .fillna(0)
)

df_sum_like_ratio
```

## 11- For each (trending_date, country) pair, the video with the largest number of views

```python
# Sum views for each (trending_date, country) pair
top_trending_videos = df.groupby(["trending_date", "country"])["views"].sum()

# Turn the groupby result back into a DataFrame
top_trending_videos = top_trending_videos.reset_index()

# Sort by total views in descending order
top_trending_videos = top_trending_videos.sort_values("views", ascending=False)

top_trending_videos
```

## 12- Divide trending_date into three columns: year, month, day

```python
# Convert trending_date (string) into a proper datetime;
df["trending_datetime"] = pd.to_datetime(
    df["trending_date"],
    format="%y.%d.%m",   # year.day.month
    errors="coerce"
)

# Extract year, month and day into separate columns
df["trend_year"] = df["trending_datetime"].dt.year
df["trend_month"] = df["trending_datetime"].dt.month
df["trend_day"] = df["trending_datetime"].dt.day


df[["video_id", "trending_date", "trend_year", "trend_month", "trend_day"]]
```

## 13- For each (month, country) pair, the video with the largest number of views

```python
# Aggregate total views per (trend_month, country, video_id)
top_trending_videos_month = df.groupby(
    ["trend_month", "country", "video_id"]
)["views"].sum()

# Convert groupby result to a DataFrame
top_trending_videos_month = top_trending_videos_month.reset_index()

# Sort by views to see the videos with most views per month/country
top_trending_videos_month = top_trending_videos_month.sort_values(
    "views", ascending=False
)

top_trending_videos_month
```

## 14- Read all json files with the video categories

```python
import json

# Re-use the same data_path and country_list defined above

all_categories = []

for country in country_list:
    # e.g. "US_category_id.json"
    jf = data_path / f"{country}_category_id.json"
    
    # Read file content as text 
    raw_text = jf.read_text(encoding="utf-8")
    
    # Parse JSON into a Python dictionary
    data = json.loads(raw_text)
    
    # Extract the "items" list that contains category metadata
    items = data.get("items", [])
    
    # Flatten nested JSON into a DataFrame
    cat_df = pd.json_normalize(items)
    
    # Keep only ID, title, and assignable flag, and rename columns
    cat_df = cat_df[["id", "snippet.title", "snippet.assignable"]].rename(
        columns={
            "id": "category_id",
            "snippet.title": "category_name",
            "snippet.assignable": "assignable"
        }
    )

    # Make sure category_id is a string and add the country column
    cat_df["category_id"] = cat_df["category_id"].astype(str)
    cat_df["country"] = country   

    # Store this intermediate DataFrame
    all_categories.append(cat_df)

# One big DataFrame with all categories and all countries
categories_df = pd.concat(all_categories, ignore_index=True)

categories_df.head(30)
```

## 15. For each country, determine how many videos have a category that is not assignable.

```python
# Filter categories that are NOT assignable
categories_nassignable = categories_df[categories_df["assignable"] == False]

# Ensure the join keys have the same type (string) in both dataframes
categories_nassignable["category_id"] = categories_nassignable["category_id"].astype(str)
df["category_id"] = df["category_id"].astype(str)

# Inner join: keep only rows whose (category_id, country) exist in both tables
df_merge = categories_nassignable.merge(
    df,
    on=["category_id", "country"],
    how="inner"
)

# Group by country and non-assignable category_id and count how many videos
assignable_video = df_merge.groupby(["country", "category_id"])[["video_id"]].count()
assignable_video=assignable_video.reset_index()
assignable_video
```
