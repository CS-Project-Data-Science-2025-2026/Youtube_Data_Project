## 1- Create a single dataframe with the concatenation of all input csv files, adding a column called country

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Path where all the country CSV files are stored
data_path = Path("/Users/tonydawra/Downloads/projectfocstry/datasets")

# List of country codes used in the file names (e.g. "USvideos.csv")
country_list = ['CA','DE','FR','GB','IN','JP','KR','MX','RU','US']

# This list will temporarily store each country's DataFrame
all_files = []

# Loop over every country code, read its CSV and add a "country" column
for country in country_list:
    # Build the full CSV path for the current country
    file_path = data_path / f"{country}videos.csv"
    
    # Read the CSV; ignore errors in the encoding if any character is problematic
    df_country = pd.read_csv(file_path, encoding='utf-8', encoding_errors="ignore")
    
    # Add the ISO country code so that later we know where each row comes from
    df_country["country"] = country
    
    # Store this temporary DataFrame in the list
    all_files.append(df_country)

# Concatenate all country DataFrames into a single big DataFrame
df = pd.concat(all_files, ignore_index=True)
```

## 2- Delete no tags

```python
# Keep only rows where the "tags" column is different from the literal string "[none]"
df = df[df["tags"] != "[none]"]
df.head()
```

## 3. For each channel, determine the total number of views

```python
# Group by channel title and sum the views across all its videos
df.groupby("channel_title")[["views"]].sum()
```

## 4-Save all rows with disabled comments and disabled ratings, or that have video_error_or_removed in a new dataframe called excluded, and remove those rows from the original dataframe.

```python
excluded = df[((df["comments_disabled"] == True) & (df["ratings_disabled"] == True)) | (df["video_error_or_removed"] == True)]
excluded
```

```python
df = df[((df["comments_disabled"] != True) & (df["ratings_disabled"] != True)) | (df["video_error_or_removed"] != True)]
df
```

## 5- Add a like_ratio column storing the ratio between the number of likes and of dislikes

```python
# Create a new column with the ratio of likes to dislikes
df["like_ratio"] = df["likes"] / df["dislikes"]
df.head()
```

```python
df[["likes", "dislikes", "like_ratio"]].head()
```

## 6- Cluster the publish time into 10-minute intervals (e.g. from 02:20 to 02:30)

```python
df["publish_time"] = pd.to_datetime(df["publish_time"], errors= "coerce")
df["publish_interval_10m"] = df["publish_time"].dt.round("10min")
df[["publish_time", "publish_interval_10m"]].head()
```

## 7- For each interval, determine the number of videos, average number of likes and of dislikes.

```python
grouped = df.groupby("publish_interval_10m")

interval_stats = grouped[[ "likes", "dislikes"]].mean()
interval_stats["n_videos"] = grouped.size()
interval_stats = interval_stats.reset_index()
interval_stats
```
