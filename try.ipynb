```python
import pandas as pd
import numpy as np
from pathlib import Path
import json

data_path = Path("/Users/tonydawra/Documents/GitHub/Youtube_Data_Project/datasets")  # folder with all CSVs and JSONs
```

```python
csv_files = list(data_path.glob("*videos.csv"))
csv_files
```

## 1- Create a single dataframe with the concatenation of all input csv files, adding a column called cou

```python
dfs = []
for f in csv_files:
    country = f.stem[:2].upper()   # e.g. CAvideos.csv → "CA"
    
    # Read as UTF-8, but ignore invalid bytes
    with open(f, "r", encoding="utf-8", errors="ignore") as fh:
        tmp = pd.read_csv(fh)
    
    tmp["country"] = country
    dfs.append(tmp)

df = pd.concat(dfs, ignore_index=True)
df
```

## 2- Delete no tags

```python
df = df[df["tags"] != "[none]"]
df.head()
```

## 3. For each channel, determine the total number of views

```python
df.groupby("channel_title")["views"].sum()
```
