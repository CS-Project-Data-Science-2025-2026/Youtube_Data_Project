## 1- Create a single dataframe with the concatenation of all input csv files, adding a column called country

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Path for connect to CSV files location
data_path = Path("/Users/tonydawra/Downloads/projectfocstry/datasets")

#Abbreviations for each file according to file names
country_list = ['CA','DE','FR','GB','IN','JP','KR','MX','RU','US']

# This list for combine each country's DataFrame
all_files = []

#Loop for reading every CSV file based on country code and add a "country" column
for country in country_list:
    # Reaching each file based on country name
    file_path = data_path / f"{country}videos.csv"

    # Read the CSV; ignore errors because of "RUvideos", "JPvideos","USvideos" files
    df_country = pd.read_csv(file_path, encoding='utf-8', encoding_errors="ignore")

    # Creating new column for country names in order to concatenate
    df_country["country"] = country

    # Store this temporary DataFrame in the list
    all_files.append(df_country)

# Concatenate all DataFrames by using country column
df = pd.concat(all_files, ignore_index=True)
```

## 5- Add a like_ratio column storing the ratio between the number of likes and of dislikes

```python
#if one of the value is nan, delete that row (if there are any 0, it create error because system can't divide with 0)
df_filtered = df[['likes', 'dislikes']].dropna()

#exclude 0 values for dislike for right dividing
df_filtered = df_filtered[df_filtered['dislikes'] > 0]

#find the ratio
df_filtered['like_ratio'] = df_filtered['likes'] / df_filtered['dislikes']

#create new column in df , still keeping nan value rows ??
df.loc[df_filtered.index, 'like_ratio'] = df_filtered['like_ratio']
df
```

## 6- Cluster the publish time into 10-minute intervals (e.g. from 02:20 to 02:30)

```python
df = df.copy()
df["publish_time"] = pd.to_datetime(df["publish_time"], errors="coerce")

# start of the 10-minute interval and convert to closest 10-minute
interval_start = df["publish_time"].dt.floor("10min")

# end = start + 10 minutes
interval_end = interval_start + pd.Timedelta(minutes=10)

# string label with just hour and minute information
df["publish_interval_10m_time"] = (
    interval_start.dt.strftime("%H:%M")
    + " to "
    + interval_end.dt.strftime("%H:%M")
)

# checking df
df[["publish_time", "publish_interval_10m_time"]].head()
```

## 7- For each interval, determine the number of videos, average number of likes and of dislikes.

```python
# Grouping the videos by the 10-minute interval
interval_stats = (
    df.groupby("publish_interval_10m_time")
      .agg(
          videos_per_interval = ("video_id", "size"),   # how many videos
          avg_likes          = ("likes", "mean"),       # avg likes
          avg_dislikes       = ("dislikes", "mean")     # avg dislikes
      )
      .reset_index()
      .sort_values("publish_interval_10m_time")         # optional just to order nicely
)

interval_stats
```

## 8- For each tag, determine the number of videos
Notice that tags contains a string with several tags.

```python
# Work on a copy to keep original data
df_tags = df.copy()

# Explode the "tags" string into one row per tag:
# Drop NaN value rows for tags
# Split the "tags" string on "|" into a list
# explode() creates one row per element of that list

#exploding each tag in the cell and adding extra rows for each different tag of  each video_id
df_tags_exploded = (
    df_tags
    .dropna(subset=["tags"])
    .assign(tag_list=lambda x: x["tags"].str.split("|"))
    .explode("tag_list")
)

# Clean the exploded tag: remove extra spaces and surrounding quotes
df_tags_exploded["tag"] = (
    df_tags_exploded["tag_list"]
    .str.replace('"','')
    .str.strip("|")
)

# We no longer need the intermediate list column
df_tags_exploded = df_tags_exploded.drop(columns=["tag_list"])

# Inspect how tags look now
df_tags_exploded[["video_id", "tags", "tag", "country"]]
```

## 9- Find the tags with the largest number of videos

```python
# Count how many *distinct* videos contain each tag
videos_per_tag = (
    df_tags_exploded.groupby("tag")["video_id"]
    .nunique()                     # number of distinct videos per tag
    .reset_index(name="n_videos")
)
```

```python
# Descending number of videos and display top 10
top_tags = videos_per_tag.sort_values("n_videos", ascending=False)
top_tags.head(10)
```

## 10- For each (tag, country) pair, compute average ratio likes/dislikes


```python
# Removes the videos which like ratio values is nan (using df_tags_exploded )
valid_ratio_tags = df_tags_exploded.dropna(subset=['like_ratio'])


# Group by tag and country, then calculate the MEAN of the like_ratio

avg_like_ratio = valid_ratio_tags.groupby(['tag', 'country'])['like_ratio'].mean().reset_index()


# Assign new name to like raito column and sort
avg_like_ratio = avg_like_ratio.rename(columns={'like_ratio': 'average_like_ratio'})
avg_like_ratio = avg_like_ratio.sort_values('average_like_ratio', ascending=False)
# Display result
avg_like_ratio
```

## 14- Read all json files with the video categories

```python
import json

# Re-use the same data_path and country_list as first question

all_categories = []

for country in country_list:
    # finding each json file
    jf = data_path / f"{country}_category_id.json"

    # Read json.file content as string type
    raw_text = jf.read_text(encoding="utf-8")

    # transform the string to a Python object
    data = json.loads(raw_text)


    #items parts of the youtube API FORMAT
    #get each item, if they dont have return list
    items = data.get("items", [])

    # Flatten nested JSON into a DataFrame
    cat_df = pd.json_normalize(items)

    # Keep only ID, title, and assignable flag, and rename columns
    cat_df = cat_df[["id", "snippet.title", "snippet.assignable"]].rename(
        columns={
            "id": "category_id",
            "snippet.title": "category_name",
            "snippet.assignable": "assignable"
        }
    )

    # Make sure category_id is a string and add the country column
    cat_df["category_id"] = cat_df["category_id"].astype(str)
    cat_df["country"] = country

    # Store this intermediate DataFrame
    all_categories.append(cat_df)

# One big DataFrame with all categories and all countries
categories_df = pd.concat(all_categories, ignore_index=True)

categories_df.head(30)
```

## 15- For each country, determine how many videos have a category that is not assignable.

```python
# Filter categories that are NOT assignable
categories_nassignable = categories_df[categories_df["assignable"] == False]

# make sure category_id has same datatype in different df for merging
categories_nassignable["category_id"] = categories_nassignable["category_id"].astype(str)
df["category_id"] = df["category_id"].astype(str)


# merging 2 different df on category id and country (both df have same column name)
df_merge = categories_nassignable.merge(
    df,
    on=["category_id", "country"],
    how="inner"
)

# Group by country and non-assignable category_id and count how many videos
assignable_video = df_merge.groupby(["country", "category_id"])["video_id"].count()

assignable_video
```
